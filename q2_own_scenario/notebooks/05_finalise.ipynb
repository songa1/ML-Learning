How this works: split the cleaned data, fit preprocessing on train only, save encoded splits and the fitted transformer, and verify that leakage columns are absent from model inputs.

```python
import sys
from pathlib import Path
ROOT = Path.cwd()
while not (ROOT / 'src').exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import joblib
from src.q2_config import RAW_FILE, PROCESSED_DIR, TARGET, LEAKAGE_COLUMNS
from src.q2_pipeline import load_raw, prepare_dataset
cleaned, train, validation, test, encoded, labels, transformer = prepare_dataset(load_raw(RAW_FILE))
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)
for name, data in encoded.items():
    output = data.copy()
    output[TARGET] = labels[name][TARGET].to_numpy()
    output['high_performer'] = labels[name]['high_performer'].to_numpy()
    output.to_csv(PROCESSED_DIR / f'q2_{name}_model_ready.csv', index=False)
joblib.dump(transformer, PROCESSED_DIR / 'q2_preprocessor.joblib')
print('split sizes:', len(train), len(validation), len(test))
print('train-only high performer threshold:', train[TARGET].quantile(1 / 3))
print('leakage columns excluded:', LEAKAGE_COLUMNS)
print('saved model-ready files and preprocessor')
```
